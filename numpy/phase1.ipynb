## Numpy array and basics 

```python
import numpy as np
```

### creating array from list

```python
arr = np.array([1, 2, 3, 4, 5])
print("1D array:", arr)

arr2D = np.array([[1, 2, 3], [4, 5, 6]])
print("2D array:", arr2D)
```

### list vs numpy

```python
py_list =[1,2,3] 
print("python list multiplication: ", py_list*2)

np_arr = np.array([1,2,3]) #element wise multiplication
print("python list multiplication: ", np_arr*2)

import time

start = time.time()
py_list = [i*2 for i in range (1000000)] # creates number array ranging from 0 - 1000000 * 2
print("python list operation time: ", time.time() - start)

start = time.time()
np_arr = np.arange(1000000) * 2  # creates numpy array ranging from 0 to 999999, then multiplies each by 2
print("Numpy list operation time: ", time.time() - start)
```
